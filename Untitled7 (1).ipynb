```python

# HOSPITAL APPOINTMENT MANAGER

# Empty lists
patients = []
appointments = []

# Predefined doctors
doctors = [
    ["D101", "Dr. Amit", "Cardiologist"],
    ["D102", "Dr. Priya", "Dermatologist"],
    ["D103", "Dr. Rahul", "Neurologist"]
]

while True:

    print("\n===== HOSPITAL APPOINTMENT MANAGER =====")
    print("1. Register Patient")
    print("2. View Doctors")
    print("3. Book Appointment")
    print("4. View Appointments")
    print("5. Exit")

    choice = input("Enter your choice: ")

    # 1. Register Patient
    if choice == "1":

        print("\n--- Patient Registration ---")

        patient_id = input("Enter Patient ID: ")
        name = input("Enter Patient Name: ")
        age = input("Enter Age: ")

        patients.append([patient_id, name, age])

        print("Patient registered successfully!")

    # 2. View Doctors
    elif choice == "2":

        print("\n--- Available Doctors ---")

        for doctor in doctors:
            print("Doctor ID:", doctor[0])
            print("Name:", doctor[1])
            print("Specialization:", doctor[2])
            print("----------------")

    # 3. Book Appointment
    elif choice == "3":

        print("\n--- Book Appointment ---")

        patient_id = input("Enter Patient ID: ")

        patient_found = False

        for patient in patients:
            if patient[0] == patient_id:
                patient_found = True

        if patient_found == False:
            print("Patient not registered.")

        else:

            for doctor in doctors:
                print(doctor[0], doctor[1], doctor[2])

            doctor_id = input("Enter Doctor ID: ")

            doctor_found = False

            for doctor in doctors:
                if doctor[0] == doctor_id:
                    doctor_found = True

            if doctor_found == False:
                print("Doctor not found.")

            else:

                date = input("Enter appointment date: ")
                time = input("Enter appointment time: ")

                appointments.append(
                    [patient_id, doctor_id, date, time]
                )

                print("Appointment booked successfully!")

    # 4. View Appointments
    elif choice == "4":

        print("\n--- All Appointments ---")

        if len(appointments) == 0:
            print("No appointments booked.")

        else:
            for appointment in appointments:
                print("Patient ID:", appointment[0])
                print("Doctor ID:", appointment[1])
                print("Date:", appointment[2])
                print("Time:", appointment[3])
                print("----------------")

    # 5. Exit
    elif choice == "5":

        print("Thank you for using the system!")
        break

    else:
        print("Invalid choice. Try again.")
```
